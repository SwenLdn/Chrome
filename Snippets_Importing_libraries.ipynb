```
import pandas as pd
import re
from reportlab.lib.pagesizes import letter, landscape
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle, Paragraph
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle

# Raw pip log input
pip_log = """Requirement already satisfied: matplotlib-venn in /usr/local/lib/python3.12/dist-packages (1.1.2)
Requirement already satisfied: matplotlib in /usr/local/lib/python3.12/dist-packages (from matplotlib-venn) (3.10.0)
Requirement already satisfied: numpy in /usr/local/lib/python3.12/dist-packages (from matplotlib-venn) (2.0.2)
Requirement already satisfied: scipy in /usr/local/lib/python3.12/dist-packages (from matplotlib-venn) (1.16.3)
Requirement already satisfied: contourpy>=1.0.1 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (1.3.3)
Requirement already satisfied: cycler>=0.10 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (0.12.1)
Requirement already satisfied: fonttools>=4.22.0 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (4.63.0)
Requirement already satisfied: kiwisolver>=1.3.1 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (1.5.0)
Requirement already satisfied: packaging>=20.0 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (26.2)
Requirement already satisfied: pillow>=8 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (11.3.0)
Requirement already satisfied: pyparsing>=2.3.1 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (3.3.2)
Requirement already satisfied: python-dateutil>=2.7 in /usr/local/lib/python3.12/dist-packages (from matplotlib->matplotlib-venn) (2.9.0.post0)
Requirement already satisfied: six>=1.5 in /usr/local/lib/python3.12/dist-packages (from python-dateutil>=2.7->matplotlib->matplotlib-venn) (1.17.0)"""

# 1. Parsing process into DataFrame
parsed_data = []
pattern = r"Requirement already satisfied:\s+(?P<package>[^\s]+)\s+in\s+(?P<path>[^\s]+)(?:\s+\(from\s+(?P<dependency>[^)]+)\))?\s+\((?P<version>[^)]+)\)"

for line in pip_log.strip().split("\n"):
    match = re.search(pattern, line)
    if match:
        data = match.groupdict()
        parsed_data.append({
            "Package Spec": data["package"],
            "Installed Version": data["version"],
            "Dependency Chain": data["dependency"] if data["dependency"] else "Direct Dependency",
            "Installation Path": data["path"]
        })

df_requirements = pd.DataFrame(parsed_data)

# 2. Export to PDF Function
def export_requirements_pdf(df: pd.DataFrame, filename: str = "python_environment_requirements.pdf"):
    doc = SimpleDocTemplate(
        filename,
        pagesize=landscape(letter),
        rightMargin=36,
        leftMargin=36,
        topMargin=36,
        bottomMargin=36
    )

    elements = []
    styles = getSampleStyleSheet()

    title_style = ParagraphStyle(
        'DocTitle',
        parent=styles['Heading1'],
        fontSize=16,
        spaceAfter=14,
        textColor=colors.HexColor('#1F4E78')
    )
    elements.append(Paragraph("Python Environment Requirements Audit", title_style))

    cell_style = ParagraphStyle('Cell', parent=styles['Normal'], fontSize=8, leading=10)
    header_style = ParagraphStyle('Header', parent=styles['Normal'], fontSize=9, leading=11, textColor=colors.white)

    table_data = [[Paragraph(str(col), header_style) for col in df.columns]]
    for _, row in df.iterrows():
        table_data.append([Paragraph(str(val), cell_style) for val in row])

    pdf_table = Table(table_data, repeatRows=1, colWidths=[130, 90, 220, 280])
    pdf_table.setStyle(TableStyle([
        ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#1F4E78')),
        ('VALIGN', (0, 0), (-1, -1), 'MIDDLE'),
        ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#E5E7EB')),
        ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.HexColor('#F9FAFB'), colors.white]),
        ('TOPPADDING', (0, 0), (-1, -1), 5),
        ('BOTTOMPADDING', (0, 0), (-1, -1), 5),
    ]))

    elements.append(pdf_table)
    doc.build(elements)

# Execute export
export_requirements_pdf(df_requirements)

# Display DataFrame in console
print(df_requirements.to_string(index=False))
```
